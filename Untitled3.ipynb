```python
# logistic regreesion classifier
import numpy as np
#dataset
x = np.array([1, 2, 3, 4])# study hours
y = np.array([0, 0, 1, 1])# 0 = fail, 1 = pass
#Initial weights and learning rate
w0, w1 = 0.0, 0.0
alpha = 0.1
# Training using gradient descent
for i in range(2000):
    # linear score
  z = w0 + w1 * x
    # sigmoid
  p = 1 / (1 + np.exp(-z))
    # error
  error = p - y
  w0 = w0 - alpha * np.mean(error)
  w1 = w1 - alpha * np.mean(error * x)
    # display weights
print("w0 =", round(w0, 4))
print("w1 =", round(w1, 4))
# prediction for new student
hours = 2.5
p = 1 / (1 + np.exp(-(w0 + w1 * hours)))
print("Probability of passing:", round(p, 4 ))
print("prediction:", "Pass" if p > 0.5 
      else "Fail")
```

```python
#unsupervised clustering k-means
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
#student data
df=pd.DataFrame({
    "study_hours":[1,2,3,4,5,6],
    "attendence":[40,45,50,80,80.5,90]
})
#input features
X = df[["study_hours","attendence"]]
#create K-Means model with 2 clusters
model = KMeans(n_clusters=2,random_state=42,n_init=10)
#train the model
model.fit(X)
#get cluster labels
df["cluster"]= model.labels_
print(df)
#display cluster centers
print("\nCentroids:")
print(model.cluster_centers_)
#Visualise clusters
plt.scatter(
    df["study_hours"],
    df["attendence"],
    c=df["cluster"],
    s=100
    )
plt.xlabel("Study HOurs")
plt.ylabel("Attendence")
plt.title("K-Means Clustering")
plt.show()
```

```python
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt
#customer data
X = np.array([ 
    [20, 30],
    [22, 28], 
    [25, 35], 
    [70, 80], 
    [75, 85],
    [78, 82]
])
linked = linkage(X, method = 'ward')
dendrogram(linked)
plt.title("Hierarchical Clustering")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
X = np.array([ [1], [2], [3],
               [8], [9], [10],
               [12], [13], [20]
])
model = DBSCAN(eps=2, min_samples=3)
model.fit(X)
labels = model.labels_
print("Cluster Labels:", labels)
plt.scatter(X[:, 0], np.zeros(len(X)), c=labels)
plt.xlabel("Data Points")
plt.title("DBSCAN Clustering")
plt.show()
```
